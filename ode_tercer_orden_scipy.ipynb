Se trata de resolver esta ecuacion  ax′′′​+ax¨+bx˙+cx=sin(t)  con scipy

```python
import numpy as np
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt

# Parámetros del sistema
a = 1.5
b = 2.0
c = 1.0

# Sistema de ecuaciones de primer orden
def rates(t, f):
    x1, x2, x3 = f

    Dx1 = x2
    Dx2 = x3
    Dx3 = np.sin(t) - a*x3 - b*x2 - c*x1

    return [Dx1, Dx2, Dx3]

# Condiciones iniciales
x1_0 = 0.0   # x(0)
x2_0 = 0.0   # x'(0)
x3_0 = 0.0   # x''(0)

f0 = [x1_0, x2_0, x3_0]

# Intervalo de tiempo
t_span = (0, 20)
t_eval = np.linspace(0, 20, 1000)

# Resolver la ecuación
sol = solve_ivp(rates, t_span, f0, t_eval=t_eval)

# Gráficas
plt.figure(figsize=(10,6))
plt.plot(sol.t, sol.y[0], label="x(t)")
plt.plot(sol.t, sol.y[1], label="x'(t)")
plt.plot(sol.t, sol.y[2], label="x''(t)")
plt.xlabel("Tiempo")
plt.ylabel("Estados")
plt.legend()
plt.grid()
plt.show()
```
