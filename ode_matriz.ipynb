```python
import numpy as np
import matplotlib.pyplot as plt

a = 1.5
b = 2.0
c = 1.0

A = np.array([
    [0, 1, 0],
    [0, 0, 1],
    [-c, -b, -a]
])

B = np.array([0, 0, 1])

def rates(t, x):
    return A @ x + B * np.sin(t)
```

```python
k1 = A @ x + B * np.sin(t)
```

```python
#Parámetros de integración
h = 0.05
t0 = 0.0
tf = 20.0
N = int((tf - t0)/h)

t = np.linspace(t0, tf, N+1)

# Condiciones iniciales
f0 = np.array([0.0, 0.0, 0.0])
```

```python
#Método de Euler
f_euler = np.zeros((N+1, 3))
f_euler[0] = f0

for n in range(N):
    f_euler[n+1] = f_euler[n] + h * rates(t[n], f_euler[n])
```

```python
# Comparación gráfica (solo x(t))
plt.figure(figsize=(10,6))
plt.plot(t, f_euler[:,0], label="Euler", linestyle="--")
plt.ylabel("x(t)")
plt.legend()
plt.grid()
plt.xlabel("Tiempo")
plt.show()
```
